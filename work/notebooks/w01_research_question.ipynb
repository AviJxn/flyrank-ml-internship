## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# ## 2. The Question: Decision, Action, Cost of a Wrong Call

# * **The Core Research Question**:
# Can observable pre-decision search signals (historical impression volume, query breadth, position stability, content age) accurately rank declining pages for an editorial refresh better than static heuristic rules without leaking label targets?
# * **Unit of Analysis**: A pseudonymized content URL (`content_hash_id`) for an individual client (`client_hash_id`) over rolling historical windows (60/90 days).
# * **Target Output**: A calibrated prioritization score ($[0, 1]$ probability ranking) representing the likelihood and severity of negative traffic trajectory.
# * **The Operational Action**: Content operations and SEO teams ingest the top 50 ranked pages weekly to schedule copy updates, snippet optimizations, or structural refreshes.
# * **Cost of a False Positive (Type I Error)**: Wasted editorial budget (~$200–$500 per article) spent rewriting a page whose traffic dip was merely seasonal or transient noise.
# * **Cost of a False Negative (Type II Error)**: Compounding lost traffic equity on core institutional pages that fall off Google's Page 1, requiring months of costly re-ranking efforts.
# * **Why Data/ML is Necessary**: Simple human rules (e.g., `age > 180 days AND impressions > 500`) capture shallow volume but exhibit high tie rates and fail on nuanced query tails. Machine learning handles multi-signal interaction and calibrates risk across diverse client domains.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import numpy as np
import pandas as pd

# Load the starter data shipped with the repository
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
decline_rate = declining_pages / total_pages

# Active visibility check
active_pages = df[df["impressions_90d"] > 0]
active_share = len(active_pages) / total_pages

# Search volume vs impressions correlation
corr_vol_imp = df["search_volume"].corr(df["impressions_90d"])

# Stale heuristic baseline check
stale_pages = (df["days_since_last_update"] >= 180) & (
    df["impressions_90d"] >= 500
)
stale_precision = (
    df[stale_pages]["trend_direction"].str.lower() == "down"
).mean()

print(f"1. Total Corpus Size: {total_pages:,} pages")
print(
    f"2. Empirical Decay Rate: {decline_rate:.2%} of pages are currently declining ({declining_pages:,}/{total_pages:,})"
)
print(
    f"3. Active Exposure Ratio: {active_share:.2%} of pages receive > 0 search impressions"
)
print(
    f"4. Volume Decoupling: Search volume to actual impression correlation is near zero ({corr_vol_imp:.4f})"
)
print(
    f"5. Heuristic Baseline Precision: Static stale rule precision is only {stale_precision:.2%}"
)
```
